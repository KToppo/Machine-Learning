```python
import numpy as np
import pandas as pd
```

```python
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.svm import SVC, SVR
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.metrics import accuracy_score, r2_score
from sklearn.ensemble import BaggingClassifier, BaggingRegressor
from sklearn.datasets import load_diabetes,load_digits
```

# Bagging

    Train using the same algorithm but with different combinations of the dataset.

# Stages in Bagging

![image](https://media.datacamp.com/legacy/v1700592080/image3_78e8da325b.png)

    Bootstrapping is the process of creating random datasets from the original data (with replacement).

# Types

1) Bagging

        Row sampling with replacement.
2) Pasting

        Row sampling without replacement.
3) Random Subspaces
   
        Column sampling with or without replacement.
4) Random Patches
   
        Both row and column sampling together with or without replacement.

# Tips

    Bagging works best with models that have low bias and high variance (like Decision Trees, SVMs, and KNN).

# Loading Data

```python
data = load_digits()
data_reg = load_diabetes()
```

```python
X_clf,y_clf = pd.DataFrame(data.data,columns=data.feature_names), pd.DataFrame(data.target,columns=['Y'])
X_train_clf,X_test_clf,y_train_clf,y_test_clf = train_test_split(X_clf,y_clf,train_size=.7,stratify=y_clf)

X_reg,y_reg = pd.DataFrame(data_reg.data,columns=data_reg.feature_names), pd.DataFrame(data_reg.target,columns=['Y'])
X_train_reg,X_test_reg,y_train_reg,y_test_reg = train_test_split(X_reg,y_reg,train_size=.7)
```

```python
X_clf.shape, X_reg.shape
```

```python
X_train_clf.shape,X_test_clf.shape, X_train_reg.shape, X_test_reg.shape
```

```python
model = DecisionTreeClassifier()
model.fit(X_train_clf,y_train_clf)
y_pred = model.predict(X_test_clf)
print('Accuracy' , accuracy_score(y_pred=y_pred,y_true=y_test_clf))
print('Cross validation', np.mean(cross_val_score(model,X_clf,y_clf,cv=10,scoring='accuracy',n_jobs=-1)))
```

```python
model = DecisionTreeRegressor()
model.fit(X_train_reg,y_train_reg)
y_pred = model.predict(X_test_reg)
print('Accuracy' , accuracy_score(y_pred=y_pred,y_true=y_test_reg))
print('Cross validation', np.mean(cross_val_score(model,X_reg,y_reg,cv=10,scoring='accuracy',n_jobs=-1)))
```

# Bagging

## Classification

```python
bagging = BaggingClassifier(estimator=DecisionTreeClassifier(),
                            n_estimators=500,
                            max_samples=200,
                            max_features=X_train_clf.shape[1],
                            bootstrap=True,
                            bootstrap_features=False,
                            oob_score=True
                            n_jobs=-1
                           )
bagging.fit(X_train_clf,np.array(y_train_clf).ravel())
```

```python
y_pred = bagging.predict(X_test_clf)
print('Accuracy' , accuracy_score(y_pred=y_pred,y_true=np.array(y_test_clf).ravel()))
print('Cross validation', np.mean(cross_val_score(bagging,X_train_clf,np.array(y_train_clf).ravel(),cv=10,scoring='accuracy',n_jobs=-1)))
```

```python
bagging.oob_score_
```

## Regression

```python
bagging = BaggingRegressor(estimator=DecisionTreeClassifier(),
                            n_estimators=500,
                            max_samples=200,
                            max_features=X_train_reg.shape[1],
                            bootstrap=True,
                            bootstrap_features=False,
                            oob_score=True,
                            n_jobs=-1
                           )
bagging.fit(X_train_reg,np.array(y_train_reg).ravel())
```

```python
y_pred = bagging.predict(X_test_reg)
print('Accuracy' , r2_score(y_pred=y_pred,y_true=np.array(y_test_reg).ravel()))
print('Cross validation', np.mean(cross_val_score(bagging,X_train_reg,np.array(y_train_reg).ravel(),cv=10,scoring='r2',n_jobs=-1)))
```

```python
bagging.oob_score_
```
