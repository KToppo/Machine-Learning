```python
import pandas as pd
import numpy as np
```

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.neighbors import KNeighborsClassifier
```

# KNN (K-Nearest Neighbors)
- Classification
    - majority count of the "K" nearest neighbors

- Regression
    - Average of the "K" nearest neighbors

## How to define "K" value
- **Heuristic Approach** (sqrt(n)) -> where n = number of observations
- **Experimentation**
    - Cross-Validation
    - GridSearchCV
    - RandomSearchCV

## Overfitting & Underfitting in KNN
- In case K is too small (e.g., 1 or 2)
    - **Overfitting**: high variance and low bias

- In case K is too big (K size covers the majority of data)
    - **Underfitting**: low variance and high bias

## Limitations of KNN
- We can't use it when:
    1. Large datasets (e.g., n=500,000, f=100)
        - training (do nothing)
        - prediction (gets too slow with bigger data)
        - **KNN is a lazy learning technique**

    2. High dimensionality data
        - Curse of dimensionality (distance concept is not reliable)

    3. Outliers
        - Sensitive to outliers

    4. Non-homogeneous feature scales
        - When data is not in the same scale

    5. Imbalanced datasets
        - where:
            - (True training dataset has significantly more instances than the false training dataset) or vice versa

    6. Inference, not prediction
       - Lack of Explicit Model
       - Computational Cost at Prediction Time
       - Limited Insight into Relationships

# Loading Dataset

```python
X,y = load_breast_cancer(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X,y,train_size=.7)
```

```python
X_train.shape, y_train.shape, X_test.shape, y_test.shape
```

```python
distance = np.sqrt(np.sum(X_train - X_test[0],axis=1)**2)
distance.shape
```

```python
sort = np.sort(np.sqrt(np.sum(X_train - X_test[0],axis=1)**2))
sort[0],sort[1], sort[2], sort[3]
```

# Creating My Own Model

```python
class My_KNN_clasification:
    def __init__(self, k=5):
        self.k = k
        self.X_train = None
        self.y_train = None

    def fit(self,X_train,y_train):
        self.X_train = X_train
        self.y_train = y_train

    def predict(self, X_test):
        pred = []
        for point in X_test:
            k_neighbors_index = []
            distance = np.sqrt(np.sum(np.square(self.X_train - point),axis=1))
            sort = np.sort(distance)
            for i in range(self.k):
                k_neighbors_index.append(np.where(distance == sort[i])[0][0])
            pred.append(self.y_train[k_neighbors_index].max())
        return pred
```

# Prediction On My Mode And Accuracy Test

```python
%%time
my_knn_clf = My_KNN_clasification(k=5)
my_knn_clf.fit(X_train,y_train)

y_pred = my_knn_clf.predict(X_test)
accuracy_score(y_pred=y_pred, y_true=y_test)
```

# Prediction On SkLearn And Accuracy Test

```python
%%time
knn_clf = KNeighborsClassifier(n_neighbors=5)
knn_clf.fit(X_train,y_train)

y_pred = knn_clf.predict(X_test)
accuracy_score(y_pred=y_pred, y_true=y_test)
```
