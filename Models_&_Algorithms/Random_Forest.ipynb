```python
import numpy as np
```

```python
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.datasets import load_diabetes, load_digits
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import accuracy_score, r2_score
```

# Random Forest
- **Why this name**
  - Forest because it's a group of decision trees
  - Random because it's a bagging technique

# Bagging VS Random Forest
- **Bagging**
  - General technique can be use any base estimater
  - Feature Selection/Feature Sampling at **Tree Level**
- **Random Forest**
  - Base model only made of Desicion Tree
  - Feature Selection/Feature Sampling at **Node Level**

# Importing Data

```python
data_clf = load_digits()
data_reg = load_diabetes()

X_train_clf,X_test_clf,y_train_clf,y_test_clf = train_test_split(data_clf.data,data_clf.target,train_size=.7,stratify=data_clf.target)
X_train_reg,X_test_reg,y_train_reg,y_test_reg = train_test_split(data_reg.data,data_reg.target,train_size=.7)
```

# Random Forest Classifier

```python
dtree_clf = DecisionTreeClassifier()
dtree_clf.fit(X_train_clf,y_train_clf)

y_pred = dtree_clf.predict(X_test_clf)
print('Accuracy', accuracy_score(y_pred=y_pred,y_true=y_test_clf))
print('Cross val score', np.mean(cross_val_score(dtree_clf,data_clf.data,data_clf.target,cv=10,n_jobs=-1,scoring='accuracy')))
```

```python
rf_clf = RandomForestClassifier(n_estimators=200,max_features=5,n_jobs=-1)
rf_clf.fit(X_train_clf,y_train_clf)

y_pred = rf_clf.predict(X_test_clf)
print('Accuracy', accuracy_score(y_pred=y_pred,y_true=y_test_clf))
print('Cross val score', np.mean(cross_val_score(rf_clf,data_clf.data,data_clf.target,cv=10,n_jobs=-1,scoring='accuracy')))
```

# Random Forest Regressor

```python
dtree_reg = DecisionTreeRegressor()
dtree_reg.fit(X_train_reg,y_train_reg)

y_pred = dtree_reg.predict(X_test_reg)
print('Accuracy', r2_score(y_pred=y_pred,y_true=y_test_reg))
print('Cross val score', np.mean(cross_val_score(dtree_reg,data_reg.data,data_reg.target,cv=10,n_jobs=-1,scoring='r2')))
```

```python
rf_reg = RandomForestRegressor(n_estimators=200,max_features=3,n_jobs=-1)
rf_reg.fit(X_train_reg,y_train_reg)

y_pred = rf_reg.predict(X_test_reg)
print('Accuracy', r2_score(y_pred=y_pred,y_true=y_test_reg))
print('Cross val score', np.mean(cross_val_score(rf_reg,data_reg.data,data_reg.target,cv=10,n_jobs=-1,scoring='r2')))
```
